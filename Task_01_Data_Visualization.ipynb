```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os

%matplotlib inline

sns.set_style("whitegrid")

print("Libraries imported successfully!")
```

```python
# Remove columns with names starting with 'Unnamed'
df = df.loc[:, ~df.columns.str.contains("^Unnamed")]

# Check updated dimensions
print("Dataset shape:", df.shape)
```

```python
# Identify year columns
year_columns = [
    col for col in df.columns
    if str(col).isdigit()
]

# Select the latest year
latest_year = year_columns[-1]

print("Latest year available:", latest_year)
```

```python
# Select country name and latest year's population
population_data = df[["Country Name", latest_year]].copy()

# Rename columns for convenience
population_data.columns = ["Country", "Population"]

# Remove missing population values
population_data = population_data.dropna()

# Convert population values to numeric
population_data["Population"] = pd.to_numeric(
    population_data["Population"],
    errors="coerce"
)

# Remove any remaining missing values
population_data = population_data.dropna()

# Display first five rows
population_data.head()
```

```python
import pycountry

# Get ISO country codes
valid_country_codes = [
    country.alpha_3 for country in pycountry.countries
]

# Keep only individual countries
country_data = df[
    df["Country Code"].isin(valid_country_codes)
].copy()

# Select the latest year
country_data = country_data[["Country Name", latest_year]]

# Rename columns
country_data.columns = ["Country", "Population"]

# Remove missing values
country_data = country_data.dropna()

# Convert population to numeric
country_data["Population"] = pd.to_numeric(
    country_data["Population"],
    errors="coerce"
)

country_data = country_data.dropna()

# Display the cleaned data
country_data.head(10)
```

```python
print("Number of countries:", len(country_data))
print("Year of analysis:", latest_year)
```

```python
plt.figure(figsize=(10, 6))

plt.hist(
    country_data["Population"],
    bins=20,
    edgecolor="black"
)

plt.title("Distribution of Population Across Countries")
plt.xlabel("Population")
plt.ylabel("Number of Countries")

plt.show()
```

```python
plt.figure(figsize=(10, 6))

plt.hist(
    country_data["Population"],
    bins=20,
    edgecolor="black"
)

plt.xscale("log")

plt.title("Distribution of Country Population (Log Scale)")
plt.xlabel("Population (Logarithmic Scale)")
plt.ylabel("Number of Countries")

plt.show()
```

```python
# Extract population values
population = country_data["Population"]

# Create logarithmically spaced bins
bins = np.logspace(
    np.log10(population.min()),
    np.log10(population.max()),
    20
)

# Plot histogram
plt.figure(figsize=(12, 6))

plt.hist(
    population,
    bins=bins,
    edgecolor="black",
    alpha=0.8
)

plt.xscale("log")

plt.title("Distribution of Population Across Countries", fontsize=15)
plt.xlabel("Population (Log Scale)", fontsize=12)
plt.ylabel("Number of Countries", fontsize=12)

plt.grid(axis="y", linestyle="--", alpha=0.5)

plt.tight_layout()
plt.show()
```

```python
# Sort countries by population in descending order
top_10 = country_data.sort_values(
    by="Population",
    ascending=False
).head(10)

top_10
```

```python
plt.figure(figsize=(12, 7))

plt.barh(
    top_10["Country"],
    top_10["Population"] / 1_000_000,
    edgecolor="black"
)

plt.gca().invert_yaxis()

plt.title(
    f"Top 10 Most Populous Countries ({latest_year})",
    fontsize=15
)

plt.xlabel("Population (Millions)")
plt.ylabel("Country")

plt.grid(axis="x", linestyle="--", alpha=0.5)

plt.tight_layout()
plt.show()
```

## Analysis of the Bar Chart

The bar chart represents the population of the ten most populous countries in the selected year.

- The countries at the top of the ranking have significantly larger populations than the other countries.
- Population differences between countries are clearly visible through the lengths of the bars.
- The chart makes it easy to compare the population sizes of the selected countries.
- The visualization highlights the unequal distribution of population among countries worldwide.

## Analysis of the Histogram

The histogram illustrates the distribution of population across different countries.

- The population distribution is positively skewed (right-skewed).
- Most countries have comparatively smaller populations.
- A small number of countries have extremely large populations.
- The logarithmic scale helps visualize population values across a wide range.

The histogram helps us understand the overall population distribution and identify differences in population sizes among countries.
